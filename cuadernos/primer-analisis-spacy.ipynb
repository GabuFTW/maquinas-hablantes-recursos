## 2. Analizar una oración

```python
import spacy

nlp = spacy.load("es_core_news_sm")
texto = (
    "Los alumnos de la Complutense presentaron ayer en Madrid "
    "un corpus de cartas del siglo XVIII."
)
doc = nlp(texto)

for token in doc:
    print(token.text, token.lemma_, token.pos_, token.dep_, token.head.text, sep="\t")

for entidad in doc.ents:
    print(entidad.text, entidad.label_)
```

Cada fila muestra la forma, el lema, la categoría, la dependencia y el núcleo. Comprueba los casos ambiguos y los nombres propios con el texto de origen.

## 3. Dibujar el árbol de dependencias

```python
from spacy import displacy
from IPython.display import display, HTML

svg = displacy.render(doc, style="dep")
display(HTML(svg))
```
