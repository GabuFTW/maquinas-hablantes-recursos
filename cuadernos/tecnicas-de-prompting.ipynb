## 2. Introduce la clave sin mostrarla

La entrada se oculta. La clave solo se guarda en la variable de la sesión actual; no la escribas en una celda de código ni en un archivo que vayas a compartir.

```python
from getpass import getpass
import anthropic

api_key = getpass("Clave de API de Anthropic: ")
cliente = anthropic.Anthropic(api_key=api_key)
```

## 3. Define las instrucciones y el mensaje

```python
system_prompt = (
    "Eres corrector de estilo de una editorial de libros de texto. "
    "Los lectores tienen 12 años."
)
texto = "El alumnado deberán entregar el trabajo cuando puedan."
mensaje = f"<texto>{texto}</texto>\n\nCorrige el texto anterior."

respuesta = cliente.messages.create(
    model="claude-sonnet-5-5",
    max_tokens=300,
    system=system_prompt,
    messages=[{"role": "user", "content": mensaje}],
)
print(respuesta.content[0].text)
```
