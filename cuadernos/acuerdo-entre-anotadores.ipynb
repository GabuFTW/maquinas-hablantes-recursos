## 2. Calcular kappa de Cohen

```python
from sklearn.metrics import cohen_kappa_score, confusion_matrix

ana = ["pos"] * 25 + ["neg"] * 25
bea = ["pos"] * 20 + ["neg"] * 5 + ["pos"] * 10 + ["neg"] * 15

kappa = cohen_kappa_score(ana, bea)
print("Kappa:", kappa)
print("Coincidencias observadas:", sum(a == b for a, b in zip(ana, bea)), "de", len(ana))
print("Matriz de confusión:\n", confusion_matrix(ana, bea, labels=["pos", "neg"]))
```
