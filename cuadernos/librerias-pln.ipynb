## 1. NLTK: tokenizar y aplicar un stemmer

Esta celda instala NLTK en el entorno temporal de Colab.

```python
!pip -q install nltk
```

```python
import nltk
from nltk.stem.snowball import SnowballStemmer

nltk.download("punkt_tab")

texto = "La nueva ley de vivienda no convence a los inquilinos de Madrid."
tokens = nltk.word_tokenize(texto, language="spanish")
stemmer = SnowballStemmer("spanish")

print(tokens)
print([stemmer.stem(token) for token in tokens])
```

El *stemmer* recorta formas con reglas; no devuelve necesariamente lemas. Los recursos de NLTK se descargan aparte de la biblioteca.

## 2. spaCy: cargar un modelo de español

Se instalan la biblioteca y el paquete de modelo. La descarga puede tardar un poco.

```python
!pip -q install spacy
!python -m spacy download es_core_news_sm
```

```python
import spacy

nlp = spacy.load("es_core_news_sm")
doc = nlp(texto)

for token in doc:
    print(token.text, token.lemma_, token.pos_, token.dep_, sep="\t")

print([(ent.text, ent.label_) for ent in doc.ents])
```

Revisa la salida: el paquete propone lemas, categorías, dependencias y entidades. Es una predicción del modelo, no una anotación infalible.

## 3. Transformers: una clasificación con un modelo concreto

Esta parte es opcional. Instala PyTorch y Transformers; después se descargarán los archivos del modelo de sentimiento. Puede tardar más y necesita memoria.

```python
!pip -q install transformers torch
```

```python
from transformers import pipeline

clasificador = pipeline(
    "sentiment-analysis",
    model="pysentimiento/robertuito-sentiment-analysis",
)

clasificador(texto)
```

El modelo se ajustó con tuits en español del corpus TASS 2020 y devuelve `POS`, `NEG` o `NEU` con una puntuación. Se entrenó con tuits, no con noticias: la puntuación no es una medida validada del tono de un titular político. Consulta la [ficha del modelo](https://huggingface.co/pysentimiento/robertuito-sentiment-analysis).
